ploting sin linea para trazar solos los marcadores ,pueden ser notaciones
de cadenas de acceso directo "o" que significa "anillos"
ejemplo:dibuja dos puntos en el diagrama uno en la pocicion (1,3) y otro en la pocicion (8,10)

```python
import matplotlib.pyplot as plt
import numpy as np

xpuntos = np.array([1,8])
ypuntos = np.array([3,10])

plt.plot(xpuntos,ypuntos)
plt.show()
```

programas de divercion (scater)
se emplean para mostrar numeracion numericas o agrupacio
ejemplo:

```python
import matplotlib.pyplot as plt


altura = [150,160,170,180,190,200]
pesos = [50,60,70,80,90,100]

plt.scatter(altura,pesos,color = "red" marker = "x")
plt.titled("altura vs peso")
plt.xlabel("altura (cm)")
plt.ylabel("pesos (kg)")
plt.show()
```

graficos de barras sirven para comparar categorias como variables tanto en orientacion vertical como horizontal
ejemplo: de barras verticales

```python
import matplotlib.pyplot as plt

categorias=['a','b','c','d']
valores =[4,7,1,8]
plt.bar(categorias,valores,color ="orange")
plt.title("grafico de barras")
plt.xlabel("categorias")
plt.ylabel("valores")
plt.show()
```
